### Task 14 – Missing Value Handling

```python
# Display missing values before handling
print("Missing values before handling:")
print(df.isnull().sum())

# -----------------------------------------
# Handle missing values in categorical columns
# -----------------------------------------

categorical_columns = [
    'vehicle_brand',
    'vehicle_model',
    'vehicle_type',
    'battery_manufacturer',
    'battery_chemistry',
    'drive_type',
    'fleet_or_private',
    'terrain_type'
]

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])


# -----------------------------------------
# Handle missing values in numerical columns
# -----------------------------------------

numerical_columns = df.select_dtypes(include=np.number).columns

for column in numerical_columns:
    if column != 'battery_failure':
        df[column] = df[column].fillna(df[column].median())


# -----------------------------------------
# Verify missing values after handling
# -----------------------------------------

print("\nMissing values after handling:")
print(df.isnull().sum())
```

### Task 19 – IQR-Based Outlier Detection

```python
# Outlier Detection using IQR Method

# 1. Battery Health Percentage

Q1_health = df['battery_health_percent'].quantile(0.25)
Q3_health = df['battery_health_percent'].quantile(0.75)

IQR_health = Q3_health - Q1_health

lower_bound_health = Q1_health - 1.5 * IQR_health
upper_bound_health = Q3_health + 1.5 * IQR_health

outliers_health = df[
    (df['battery_health_percent'] < lower_bound_health) |
    (df['battery_health_percent'] > upper_bound_health)
]

print("Battery Health Percentage")
print("Q1:", Q1_health)
print("Q3:", Q3_health)
print("IQR:", IQR_health)
print("Lower Bound:", lower_bound_health)
print("Upper Bound:", upper_bound_health)
print("Number of outliers:", len(outliers_health))

print("\nOutlier values:")
print(outliers_health[['vehicle_id', 'battery_health_percent']])


# 2. Cycle Count

Q1_cycle = df['cycle_count'].quantile(0.25)
Q3_cycle = df['cycle_count'].quantile(0.75)

IQR_cycle = Q3_cycle - Q1_cycle

lower_bound_cycle = Q1_cycle - 1.5 * IQR_cycle
upper_bound_cycle = Q3_cycle + 1.5 * IQR_cycle

outliers_cycle = df[
    (df['cycle_count'] < lower_bound_cycle) |
    (df['cycle_count'] > upper_bound_cycle)
]

print("\n" + "=" * 50)

print("Cycle Count")
print("Q1:", Q1_cycle)
print("Q3:", Q3_cycle)
print("IQR:", IQR_cycle)
print("Lower Bound:", lower_bound_cycle)
print("Upper Bound:", upper_bound_cycle)
print("Number of outliers:", len(outliers_cycle))

print("\nOutlier values:")
print(outliers_cycle[['vehicle_id', 'cycle_count']])


# 3. Battery Stress Index

Q1_stress = df['battery_stress_index'].quantile(0.25)
Q3_stress = df['battery_stress_index'].quantile(0.75)

IQR_stress = Q3_stress - Q1_stress

lower_bound_stress = Q1_stress - 1.5 * IQR_stress
upper_bound_stress = Q3_stress + 1.5 * IQR_stress

outliers_stress = df[
    (df['battery_stress_index'] < lower_bound_stress) |
    (df['battery_stress_index'] > upper_bound_stress)
]

print("\n" + "=" * 50)

print("Battery Stress Index")
print("Q1:", Q1_stress)
print("Q3:", Q3_stress)
print("IQR:", IQR_stress)
print("Lower Bound:", lower_bound_stress)
print("Upper Bound:", upper_bound_stress)
print("Number of outliers:", len(outliers_stress))

print("\nOutlier values:")
print(outliers_stress[['vehicle_id', 'battery_stress_index']])
```

### Task 25 – Categorical Encoding

```python
from sklearn.preprocessing import LabelEncoder

# ==========================================
# 1. Label Encoding
# ==========================================

# Create LabelEncoder
le = LabelEncoder()

# Example: Label Encoding Vehicle Type
df['vehicle_type_encoded'] = le.fit_transform(
    df['vehicle_type']
)

print("Vehicle Type Encoding:")
print(
    df[['vehicle_type', 'vehicle_type_encoded']]
    .drop_duplicates()
    .sort_values('vehicle_type_encoded')
)


# ==========================================
# 2. One-Hot Encoding
# ==========================================

# Nominal categorical variables
categorical_columns = [
    'vehicle_brand',
    'vehicle_model',
    'battery_manufacturer',
    'battery_chemistry',
    'drive_type',
    'fleet_or_private',
    'terrain_type'
]

# Apply One-Hot Encoding
df = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=True
)

print("\nDataFrame after One-Hot Encoding:")
display(df.head())

print("\nNew Column Names:")
print(df.columns)
```

### Task 26 – Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler

# Numerical columns suitable for scaling
numeric_columns = [
    'battery_capacity_kwh',
    'odometer_km',
    'cycle_count',
    'battery_health_percent',
    'battery_stress_index'
]

# Make a copy of the numerical data
numeric_data = df[numeric_columns].copy()

# Handle missing values before scaling
numeric_data = numeric_data.fillna(
    numeric_data.median()
)


# ==========================================
# Standard Scaling
# ==========================================

standard_scaler = StandardScaler()

standard_scaled = standard_scaler.fit_transform(
    numeric_data
)

df_standard = numeric_data.copy()

df_standard[numeric_columns] = standard_scaled

print("Standard Scaled Data:")
display(df_standard.head())


# ==========================================
# Min-Max Scaling
# ==========================================

minmax_scaler = MinMaxScaler()

minmax_scaled = minmax_scaler.fit_transform(
    numeric_data
)

df_minmax = numeric_data.copy()

df_minmax[numeric_columns] = minmax_scaled

print("\nMin-Max Normalized Data:")
display(df_minmax.head())
```
