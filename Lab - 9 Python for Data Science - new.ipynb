### 6. Given roll numbers 101–120, randomly select 5 students without replacement and demonstrate shuffling and permutation.

```python
roll_numbers = np.arange(101, 121)
selected = np.random.choice(roll_numbers, 5, replace=False)
np.random.shuffle(roll_numbers)
permuted = np.random.permutation(roll_numbers)
print("5 randomly selected students:")
print(selected)
print("\nAfter shuffling:")
print(roll_numbers)
print("\nRandom permutation:")
print(permuted)
```

### 10. Create an unsorted array of student marks and display it in ascending and descending order. Also find the sorted indices.

```python
marks = np.array([78, 85, 92, 67, 74, 88, 95, 81, 69, 90, 76, 84, 73, 89, 80])
aesc=np.sort(marks)
desc=np.sort(marks)[::-1]
sorted_indices=np.argsort(marks)

print("Original:",marks)
print("Aesc:",aesc)
print("Desc:",desc)
print("Sorted indices:", sorted_indices)
```

### 17. Given an array, write a program to display elements that are divisible by 5 and greater than 25, and then display elements that are either less than 20 or greater than 45.

```python
arr = np.array([10, 15, 20, 25, 30, 35, 40, 45, 50, 55])
```

```python
print("Divisible by 5 and greater than 25:")
print(arr[(arr % 5 == 0) & (arr > 25)])

print("Less than 20 or greater than 45:")
print(arr[(arr < 20) | (arr > 45)])
```
