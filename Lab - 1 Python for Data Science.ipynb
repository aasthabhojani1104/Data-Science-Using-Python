### 17. WAP to convert total seconds into hours, min and seconds

```python
tsec=int(input("Enter sec:"))
hour=int(tsec/3600)
min=int((tsec%3600)/60)
sec=int(tsec%60)
print(hour, ":" , min, ":", sec)
```

### 18. WAP to convert number of days into year, week and days.

```python
tdays=int(input("Enter days:"))
year=int(tdays/365)
remainingdays=int(tdays%365)
weeks=int(remainingdays/7)
days=int(remainingdays%7)
print(year, ",", weeks, ",", days)
```
