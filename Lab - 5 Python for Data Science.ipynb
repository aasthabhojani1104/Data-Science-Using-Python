### 14) WAP to extract elements with frequency greater than K.

```python
L = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

K = int(input("Enter K: "))

result = []

for i in L:
    if L.count(i) > K and i not in result:
        result.append(i)

print("Elements with frequency greater than K:", result)
```

### 16) WAP to create a new list (fruit whose name starts with 'b') from the list of fruits given by user.

```python
fruits = input("Enter fruits separated by space: ").split()

result = [fruit for fruit in fruits if fruit.lower().startswith('b')]

print("Fruits starting with 'b':", result)
```

### 17) WAP to create a list of common elements from given two lists.

```python
L1 = [1, 2, 3, 4, 5]
L2 = [3, 4, 5, 6, 7]

result = []

for i in L1:
    if i in L2:
        result.append(i)

print("Common elements:", result)
```
