### 7. WAP to print sum of series 1 – 2 + 3 – 4 + 5 – 6 + 7 ... n.

```python
n = int(input("Enter n: "))

sum = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        print("-", i, end=" ")
        sum -= i
    else:
        if i == 1:
            print(i, end=" ")
        else:
            print("+", i, end=" ")
        sum += i

print("\nSum =", sum)
```

### 11. WAP to find whether the given number is prime or not.

```python
n=int(input("enter n"))
count=0
for i in range(1,n+1):
    if(n%i==0):
       count+=1
if(count==2):
    print(n,"is prime")
else:
    print(n,"not prime")
```

### 12. WAP to print sum of digits of given number.

```python
n=int(input("enter n"))
sum=0
while n>0:
    digit=n%10
    sum+=digit
    n=n//10
print(sum)
```

### 13. WAP to check whether the given number is palindrome or not.

```python
n=int(input("enter n"))
temp=n
reverse=0
while n>0:
    digit=n%10
    reverse=reverse*10+digit
    n=n//10
if temp==reverse:
    print("Palindrom")
else:
    print("not palindrom")
```
