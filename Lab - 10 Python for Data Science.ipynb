### 1. Series in Pandas

Create a Pandas Series containing the marks of 8 students. Display the Series, its index, values, and data type.

```python
import pandas as pd
import numpy as np
```

```python
marks=pd.Series([80,74,88,94,90,67])
print("Series:")
print(marks)

print("Index:")
print(marks.index)

print("Values:")
print(marks.values)

print("Data Type")
print(marks.dtype)
```

### 4. Creating Time Series

Create a date range for 10 consecutive days starting from '2026-09-01' and create a Pandas Series containing daily sales values for those dates. Display the resulting time series.

```python
dates=pd.date_range(start='2026-09-01',periods=10,freq='D')
sales=pd.Series([1200, 1500, 1100, 1800, 2000, 1750, 2100, 1900, 2250, 2400],index=dates)

print("Daily Series:")
print(sales)
```

### 5. Time Series with Frequency

Create a time series for the first 6 months of 2026 using a suitable date frequency. Store monthly revenue values in the Series and display the result.

```python
dates=pd.date_range(start='2026-01-01',periods=6,freq='MS')
sales=pd.Series([1200, 1500, 1100, 1800, 2000, 1750],index=dates)

print("Monthly Revenue:")
print(sales)
```

### 8. Conditional Selection

Using the `students` DataFrame, display: i) students having Marks greater than 70, ii) students belonging to the 'CSE' department, and iii) students having Marks between 60 and 80.

```python
students = pd.DataFrame({
    "RollNo": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Karan", "Sneha"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT", "CSE"],
    "Marks": [85, 92, 68, 55, 78, 72],
    "City": ["Ahmedabad", "Surat", "Vadodara", "Rajkot", "Ahmedabad", "Surat"]
})

print("Marks > 70")
print(students[students["Marks"] > 70])

print("Students belonging to CSE department:")
print(students[students["Department"] == "CSE"])

print("Students with Marks between 60 and 80:")
print(students[(students["Marks"] >= 60) & (students["Marks"] <= 80)])

```

### 9. Multiple Conditions

Using the `students` DataFrame, select students who belong to the 'CSE' department and have Marks greater than or equal to 70. Also display students who belong to either 'CSE' or 'IT'.

```python
students = pd.DataFrame({
    "RollNo": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Karan", "Sneha"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT", "CSE"],
    "Marks": [85, 92, 68, 55, 78, 72],
    "City": ["Ahmedabad", "Surat", "Vadodara", "Rajkot", "Ahmedabad", "Surat"]
})

print("CSE students with Marks >= 70:")
print(students[(students["Department"] == "CSE") & (students["Marks"] >= 70)])
print("Students belonging to either CSE or IT:")
print(students[students["Department"].isin(["CSE", "IT"])])
```
