### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def prime(val):
    for i in range(2,val):
        if val%i==0:
            return 0
        else:
            return 1
prime(24)
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibo(n):
    if n<=1 :
        return 0
    elif n==2:
        return 1
    return fibo(n-1)+fibo(n-2)

fibo(10)
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
dict1 = {'a': 10, 'b':20, 'c':30, 'd':40}
dict2 = {}
def invert(dic):
    for k,v in dic.items():
        dict2[v]=k
    return dict2
invert(dict1)
```

### 12) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
str1 = 'AbcDEfgh'

def cnt(st):
    no_upper = no_lower = 0
    for i in st:
        if i.islower():
            no_upper += 1
        if i.isupper():
            no_lower += 1
    print("no_lower = ",no_lower," no_upper = ",no_upper) 

cnt(str1)
```

### 16) Write udfs to call the functions with following types of arguments:
1. Positional Arguments
2. Keyword Arguments
3. Default Arguments
4. Variable Legngth Positional(*args) & variable length Keyword Arguments (**kwargs)
5. Keyword-Only & Positional Only Arguments

```python
# 1. Positional Arguments

def pos_arg(a,b,c):
    print(f"a = {a} , b = {b} , c = {c}")

pos_arg(10,20,30)
```

```python
#2. Keyword Arguments

def key_arg(a,b,c):
    print(f"a = {a} , b = {b} , c = {c}")

key_arg(c=10,a=30,b=20)
```

```python
#3. Default Argument

def def_arg(a,b,c=0):
    print(f"a = {a} , b = {b} , c = {c}")

def_arg(11,2)
def_arg(1,22,3)
```

```python
#4. Positional Only Argument

def pos_only(a,b,/):
    print(f"a = {a} , b = {b}")

pos_only(20,10)
pos_only(a=10,b=20)
```

```python
#4. Keyword Only Argument

def key_only(*,a,b):
    print(f"a = {a} , b = {b}")

key_only(b=10,a=20)
key_only(20,10)
```

```python
#5. Variable Length Postional Argument

def varlen(*num):
    print("Sum of all the numbers : ",sum(num))

varlen(10,20,30)
```

```python
#5. Variable Length Postional Argument

def varlen(**num):
    print("First number : ",num["a"],"Last number : ",num["c"])

varlen(a=10,b=20,c=30)
```
